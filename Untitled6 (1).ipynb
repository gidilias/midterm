```python
import json
import pandas as pd
from pathlib import Path

folder = Path("runs/deepseek_main")

manifest = json.loads((folder / "manifest.json").read_text())
print("Статус:", manifest["status"])

rows = [
    json.loads(line)
    for line in (folder / "outputs.jsonl").read_text().splitlines()
]
print("Количество ответов:", len(rows))

summary = pd.DataFrame(
    json.loads((folder / "summary.json").read_text())
)
display(summary)
```
